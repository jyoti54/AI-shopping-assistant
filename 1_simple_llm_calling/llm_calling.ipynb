```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

response = llm.invoke("how many moons Jupiter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "How many moons does Jupiter have?"]
])
print(response.text)
```

```python
prompt="Write a single tagline for a quirkly coffee shop. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash", temperature=0)
response=llm.invoke(prompt)
print(response.text)
```

```python
llm_explore = ChatGoogleGenerativeAI(model="gemini-3.6-flash", temperature=0)

res = llm_explore.invoke(prompt)
print(res.text)
```

## Call LLM using Groq models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3.8-27b")

messages = [
    (
        "system",
        "You are a helpful assistant that gives a short answer to a question",
    ),
    (
        "human",
        "How many moons does Jupiter have?"
    ),
]

ai_message = llm.invoke(messages)
print(ai_message.content)
```
